#### Basics:
1. Getline: <br>
-> Used to get full string entered along with spaces.<br>
Normally, we do:<br>
&emsp;string name;<br>
&emsp;cout << "Enter your name: ";<br>
&emsp;cin>>name;<br>
But this won't work with full names if they have spaces, so we use:<br>
&emsp;string name;<br>
&emsp;cout << "Enter your name: ";<br>
&emsp;getline(cin, name);<br>

2. Vector:<br>
-> Used to define a dynamic array in c++.<br>
To declare: vector(int) fc;<br>
vector function to add items: vector.puch_back(item);<br>
vector function to check empty or not: fc.empty()<br>
vector function to get its size: fc.size()<br>

3. static_cast(char)variable:<br>

4. Ternary operators: <br>

5. Break and Continue:<br>
