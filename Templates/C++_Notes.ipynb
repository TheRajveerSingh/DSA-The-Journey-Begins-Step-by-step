#### Basics:
1. Getline: <br>
-> Used to get full string entered along with spaces.<br>
Normally, we do:<br>
&emsp;string name;<br>
&emsp;cout << "Enter your name: ";<br>
&emsp;cin>>name;<br>
But this won't work with full names if they have spaces, so we use:<br>
&emsp;string name;<br>
&emsp;cout << "Enter your name: ";<br>
&emsp;getline(cin, name);<br>

-> This method is useful when parameters given to functions are too large, and we don't want to waste time copyimg them, and imstead, take their value as they are.
