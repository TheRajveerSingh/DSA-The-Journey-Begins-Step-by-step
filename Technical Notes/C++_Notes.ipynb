17. Returning type for bool:<br>
-> Normally for bool, we return 'true' and 'false', but alternatively, we can return '1' and '0'.<br>
Where '1' or any number except o would return 'true', while '0' would return 'false'.<br>
However, when we call a function (bool), 'true' and 'false' won't be printed. Instead we'd get '1' and '0'.<br>
To print them, we'll have to write something like:<br>
if (function){cout<<"True";}<br>
else{cout<<"False";}<br>
Also, speacially for a bool function, we don't need to compare it with anything (ex: no need for : if(function=='true'))<br>
Automatically, whatever is returned, if its true, then that condition runs.<br>
